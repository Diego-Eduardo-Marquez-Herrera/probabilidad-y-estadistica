 el problema 1 es de tipo poisson porque estamos contando cuántas personas entran a la unidad de cuidados intensivos a lo largo de un intervalo de tiempo continuo

```python
from scipy.stats import poisson

lam = 5

p_2 = poisson.pmf(2, lam)
print(f"a) Probabilidad de 2 personas: {p_2:.4f} ({p_2 * 100:.2f}%)")

p_exceda_10 = 1 - poisson.cdf(10, lam)
print(f"b) Probabilidad de que exceda 10: {p_exceda_10:.4f} ({p_exceda_10 * 100:.2f}%)")

if p_exceda_10 < 0.05:
    print("Respuesta: NO es probable que x exceda de 10, ya que es un evento raro (menor al 5%).")
else:
    print("Respuesta: SÍ es probable que x exceda de 10.")
```

Es una distribución Binomial porque tenemos un número fijo de intentos y cada motor solo tiene dos opciones: salir defectuoso o salir bien. Aparte, la probabilidad de que un motor falle siempre se mantiene igual para todos

```python
from scipy.stats import binom

n = 1000
p = 0.001

p_0 = binom.pmf(0, n, p)
print(f"• Probabilidad de 0 defectuosos P(X = 0): {p_0:.4f} ({p_0 * 100:.2f}%)")

p_3 = binom.pmf(3, n, p)
print(f"• Probabilidad de 3 defectuosos P(X = 3): {p_3:.4f} ({p_3 * 100:.2f}%)")

p_4 = binom.pmf(4, n, p)
print(f"• Probabilidad de 4 defectuosos P(X = 4): {p_4:.4f} ({p_4 * 100:.2f}%)")
```
