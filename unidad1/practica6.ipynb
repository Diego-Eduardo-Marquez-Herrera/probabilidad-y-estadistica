```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import math
```

```python
df = pd.read_csv('autos2.csv')
```

```python
precios = df['price'].dropna()
```

```python
minimo_precio = precios.min()
maximo_precio = precios.max()
moda_precio = precios.mode()[0]  # Tomamos el primer valor modal
media_precio = precios.mean()
mediana_precio = precios.median()
desviacion_precio = precios.std()

print("--- MEDIDAS DE TENDENCIA CENTRAL Y DISPERSIÓN ---")
print(f"Precio Mínimo: ${minimo_precio:.2f}")
print(f"Precio Máximo: ${maximo_precio:.2f}")
print(f"Moda de Precios: ${moda_precio:.2f}")
print(f"Media de Precios: ${media_precio:.2f}")
print(f"Mediana de Precios: ${mediana_precio:.2f}")
print(f"Desviación Estándar: ${desviacion_precio:.2f}")

print("\n--- RESUMEN ESTADÍSTICO CON DESCRIBE() ---")
print(precios.describe())
```

```python
n = len(precios)
num_bins = int(1 + math.log2(n)) # Aplicación de la Regla de Sturges

print(f"\nNúmero total de datos (n): {n}")
print(f"Cantidad óptima de bins (Sturges): {num_bins}")

plt.figure(figsize=(8, 5))
plt.hist(precios, bins=num_bins, color='skyblue', edgecolor='black', alpha=0.8)
plt.axvline(media_precio, color='red', linestyle='--', linewidth=2, label=f'Media: ${media_precio:.2f}')
plt.axvline(mediana_precio, color='green', linestyle='-', linewidth=2, label=f'Mediana: ${mediana_precio:.2f}')

plt.title('Distribución de Precios de Automóviles')
plt.xlabel('Precio (USD)')
plt.ylabel('Número de Autos')
plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(x='body-style', y='price', data=df, palette='Set2')
plt.title('Relación entre el Precio y el Tipo de Vehículo')
plt.xlabel('Tipo de Vehículo (Body Style)')
plt.ylabel('Precio (USD)')
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='drive-wheels', y='price', data=df, palette='Pastel1')
plt.title('Relación entre el Precio y el Tipo de Tracción')
plt.xlabel('Tipo de Tracción (Drive Wheels)')
plt.ylabel('Precio (USD)')
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(x='engine-location', y='price', data=df, palette='coolwarm')
plt.title('Relación entre el Precio y la Ubicación del Motor')
plt.xlabel('Ubicación del Motor')
plt.ylabel('Precio (USD)')
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```
