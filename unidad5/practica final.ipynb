```python
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from scipy.stats import pearsonr

```

```python
df1 = pd.read_csv("grupo1.csv")
print(df1.head())
print(df1.describe())

X1 = sm.add_constant(df1["HorasEstudio"])
y1 = df1["Calificacion"]
modelo1 = sm.OLS(y1, X1).fit()

b0_1 = modelo1.params["const"]
b1_1 = modelo1.params["HorasEstudio"]
r2_1 = modelo1.rsquared
p_val_1 = modelo1.pvalues["HorasEstudio"]
r_1, _ = pearsonr(df1["HorasEstudio"], df1["Calificacion"])

print(f"\nEcuación: Calificacion = {b0_1:.4f} + ({b1_1:.4f} * HorasEstudio)")
print(f"Pendiente: {b1_1:.4f}")
print(f"Intercepto: {b0_1:.4f}")
print(f"Correlación de Pearson (r): {r_1:.4f}")
print(f"R²: {r2_1:.4f}")
print(f"Valor p: {p_val_1:.4e}")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(df1["Calificacion"], kde=True, color="skyblue", bins=10, ax=axes[0])
axes[0].set_title("Histograma de Calificaciones - Grupo 1")
axes[0].set_xlabel("Calificación Final")
axes[0].set_ylabel("Frecuencia")

sns.regplot(
    x="HorasEstudio", y="Calificacion", data=df1, ci=None, ax=axes[1],
    line_kws={"color": "red", "label": f"y = {b0_1:.2f} + ({b1_1:.2f})x"}
)
axes[1].set_title(f"Regresión Lineal - Grupo 1 (r = {r_1:.2f})")
axes[1].set_xlabel("Horas de Estudio")
axes[1].set_ylabel("Calificación Final")
axes[1].legend()

plt.tight_layout()
plt.show()
```

### Interpretación de Resultados — Grupo 1

1. **¿Explicar cuál es la interpretación del coeficiente de regresión en términos de la relación entre el tiempo de estudio y las calificaciones finales?**
   La pendiente es de -0.1061, lo que significa que por cada hora extra de estudio la calificación baja en promedio 0.11 puntos, o sea, un cambio casi nulo. El intercepto es de 85.3107 puntos cuando las horas de estudio son 0.

2. **¿Cómo se interpreta el coeficiente de correlación en términos de la fuerza y dirección de la relación entre las variables?**
   El valor de r = -0.0571 indica una relación prácticamente nula y ligeramente negativa. En la gráfica se ve que los puntos están muy dispersos sin seguir una línea clara.

3. **¿El modelo de regresión lineal es estadísticamente significativo?**
   No es estadísticamente significativo porque el valor p (0.6935) es mucho mayor a 0.05, así que no hay evidencia de una relación lineal real entre las variables en este grupo.

4. **¿Qué tan confiable es el modelo de regresión para predecir las calificaciones finales de los estudiantes basado en el tiempo de estudio?**
   No es nada confiable, ya que su $R^2$ es de 0.0033, lo que quiere decir que las horas de estudio apenas explican el 0.33% de la variación en las calificaciones.
